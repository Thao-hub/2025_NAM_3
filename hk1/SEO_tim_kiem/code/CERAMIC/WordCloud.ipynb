```python
url = 'http://localhost:5206/home/category/1/do-tho-bat-trang.html'
html = requests.get(url)
soup = BeautifulSoup(html.content)
#soup
```

```python
data = soup.select_one('div.card-body').text
```

```python
words = word_tokenize(data)
stopword = ('các', '.', ',', 'và', 'là', '-', 'có', 'với', 'cho', 'trong')
words = [word.lower() for word in words if word.lower() not in stopword]
```

```python
plt.rcParams['figure.figsize'] = (20,10)
wordcloud = WordCloud(width=1000, height=700, background_color='white')
frequencies = Counter(words)
wc = wordcloud.fit_words(frequencies=frequencies)
plt.imshow(wc)
plt.show()
```
