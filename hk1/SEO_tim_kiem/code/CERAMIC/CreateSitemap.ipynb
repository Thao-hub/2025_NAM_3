```python
from bs4 import BeautifulSoup
import requests
import xml.etree.ElementTree as et
```

```python
root = 'http://localhost:5206/'
html = requests.get(root)
soup = BeautifulSoup(html.content)
#print(html)
#soup
```

Selector (css vs javascription)
.: class
#: id
name: tag name
attribute: href, alt
a => anchor

```python
anchors = []
for div in soup.select('div.flex-fill'):
    #print(div.select_one('a').attrs['href'])
    anchors.append(div.select_one('a').attrs['href'])
anchors
```

```python
pages =[]
pageLinks = soup.select('ul.pagination a.page-link')
for link in pageLinks:
    #print(link.attrs['href'])
    pages.append(link.attrs['href'])
pages
```

```python
articles = []
for page in pages:
    sub_url = 'http://localhost:5206/' + page
    brr = getArticles(sub_url)
    articles += brr
articles
```

```python
def getArticles(sub_url):
    html = requests.get(sub_url)
    soup = BeautifulSoup(html.content)
    #print(soup)
    anchors = soup.select('h2.h6 a')
    arr = []
    for anchor in anchors:
        #print(anchor.attrs['href'])
        arr.append(anchor.attrs['href'])
    return arr
sub_url = 'http://localhost:5206/?page=2'
getArticles(sub_url)
```

```python
urlset = et.Element('urlset')
for anchor in anchors:
    url = et.SubElement(urlset, 'url')
    loc = et.SubElement(url, 'loc')
    loc.text = 'http://gombattrangtranthao.somee.com/' + anchor
for article in articles:
    url = et.SubElement(urlset, 'url')
    loc = et.SubElement(url, 'loc')
    loc.text = 'http://gombattrangtranthao.somee.com/' + article
tree = et.ElementTree(urlset)
tree.write('./WebApp/wwwroot/sitemap.xml')
    
```
