```python
from networkx import DiGraph, draw
from networkx import dijkstra_path, pagerank
import matplotlib.pyplot as plt
```

```python
g = DiGraph()
g.add_edge(1, 2)
g.add_edge(2, 3)
g.add_edge(2, 5)
g.add_edge(3, 1)
g.add_edge(4, 2)
g.add_edge(5, 4)
g.add_edge(5, 3)
draw(g, with_labels=True)
plt.show()
```

```python
dijkstra_path(g, 1, 4)
```

```python
pagerank(g)
```

```python
g2 = DiGraph()
g2.add_edge('A', 'B')
g2.add_edge('A', 'C')
g2.add_edge('A', 'D')
g2.add_edge('C', 'B')
g2.add_edge('D', 'B')
pagerank(g2)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cosine_similarity([
    [3, 7],
    [4, 6],
    [15, 21],
    [-3, 15]
])
```

```python
copus = [
    'This is the first document',
    'This document is the second document second second',
    'And this is the third one',
    'Is this the first document'
]
cv = CountVectorizer()
X = cv.fit_transform(copus).toarray()
```

```python
cv.vocabulary_
```

```python
q = ['document']
q_X = cv.transform(q).toarray()
cosine_similarity(X, q_X)
```
