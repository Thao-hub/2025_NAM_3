```python
from math import radians, sin, cos, sqrt, atan2
#tính khoảng cách
def haversine(lat1, lon1, lat2, lon2, radius=6371.0):
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)

    a = (
        sin(dlat / 2) ** 2
        + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon / 2) ** 2
    )

    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = radius * c
    return distance
```

```python
# E1
def read_cities_from_file(input_file):
    cities = []

    try:
        with open(input_file, "r", encoding="utf-8") as f:
            for line_no, line in enumerate(f, start=1):
                line = line.strip()

                # bỏ qua dòng trống
                if not line:
                    continue

                parts = [p.strip() for p in line.split(",")]
                if len(parts) != 3:
                    print(f"Dòng {line_no} sai định dạng (cần 3 phần): {line}")
                    continue

                name, lat_str, lon_str = parts

                try:
                    lat = float(lat_str)
                    lon = float(lon_str)
                except ValueError:
                    print(f"Dòng {line_no} có lat/lon không hợp lệ: {line}")
                    continue

                cities.append({"name": name, "lat": lat, "lon": lon})

        return cities

    except FileNotFoundError:
        print(f"Lỗi: Không tìm thấy file '{input_file}'.")
        return []
    except Exception as e:
        print(f"Lỗi không xác định khi đọc file: {e}")
        return []
```

```python
# E2
def write_cities_to_file(output_file, cities):
    try:
        with open(output_file, "w", encoding="utf-8") as f:
            for c in cities:
                f.write(f"{c['name']},{c['lat']},{c['lon']}\n")
        print(f"Đã ghi dữ liệu vào '{output_file}'")
    except Exception as e:
        print(f"Lỗi khi ghi file '{output_file}': {e}")
```

```python
# E3
def process_geospatial_file(input_file, output_file):
    try:
        cities = read_cities_from_file(input_file)
        if not cities:
            print("Không có dữ liệu để xử lý.")
            return

        base = cities[0]  # lấy thành phố đầu tiên làm mốc

        with open(output_file, "w", encoding="utf-8") as out:
            out.write(f"Thành phố mốc: {base['name']} ({base['lat']}, {base['lon']})\n")
            out.write("Khoảng cách (km) từ thành phố mốc đến các thành phố khác:\n")

            for c in cities[1:]:
                d = haversine(base["lat"], base["lon"], c["lat"], c["lon"])
                out.write(f"- {c['name']}: {d:.2f} km\n")

        print(f"Đã xử lý xong và ghi kết quả vào '{output_file}'")

    except Exception as e:
        print(f"Lỗi khi xử lý dữ liệu không gian: {e}")
    finally:
        print(f"Finished processing {input_file}")
```

```python
sample_data = """Tokyo,35.6895,139.6917
Los Angeles,34.0522,-118.2437
London,51.5074,-0.1278
INVALID_LINE
Sydney,-33.8688,151.2093
Paris,48.8566,2.3522
BadCoord,abc,xyz
"""

input_file = "cities.txt"
with open(input_file, "w", encoding="utf-8") as f:
    f.write(sample_data)
```

```python
cities = read_cities_from_file(input_file)
write_cities_to_file("output_cities.txt", cities)
process_geospatial_file(input_file, "geospatial_result.txt")
```
