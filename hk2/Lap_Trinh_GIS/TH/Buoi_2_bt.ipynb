```python
#E5
def parse_address(address_str):
    parts = address_str.split(", ")
    return {
        "đường": parts[0],
        "quận": parts[1],
        "thành phố": parts[2]
    }
res = parse_address("123 Phan Đình Phùng, Đà Lạt, Lâm Đồng")
print(res)
```

```python
#E6
latitude = float(lat_str[:-1])
longitude = float(lon_str[:-1])
def convert_to_wkt_point(lat, lon):
    return f"POINT({lon} {lat})"
wkt_point = convert_to_wkt_point(latitude, longitude)
print(wkt_point)
```
