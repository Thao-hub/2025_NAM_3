```python
import pandas as pd
from sklearn.linear_model import LinearRegression
df = pd.DataFrame([
    [32 ,10 ,37.9],
    [19 ,9 ,42.2],
    [13 ,5 ,47.3],
    [13 ,5 ,47.5],
    [5 ,5 ,51.5],
    [7 ,3 ,48.2],
    [34 ,7 ,40.3],
    [20 ,6 ,46.7],
    [30 ,1 ,18.8],
    [17 ,3 ,25.8],
], 
columns=['x1', 'x2', 'y'])

X = df[['x1', 'x2']]
y = df['y']

lg = LinearRegression()
lg.fit(X, y)
y_pred = lg.predict(X)

print('Predict', y_pred)
print('R^2 score', lg.score(X, y))
```
