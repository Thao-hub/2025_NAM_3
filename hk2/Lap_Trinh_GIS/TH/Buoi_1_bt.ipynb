```python
# Ex 1
cities_coords = [
    (21.0285, 105.8542),    
    # Hà Nội
    (10.8231, 106.6297),  # TP.HCM
    (16.0544, 108.2022),  # Đà Nẵng
    (12.2388, 109.1967),  # Nha Trang
    (10.0452, 105.7469),  # Cần Thơ
]
centroid_lat = sum(point[0] for point in cities_coords) / len(cities_coords)
centroid_lon = sum(point[1] for point in cities_coords) / len(cities_coords)
centroid = {
    "latitude": centroid_lat,
    "longitude": centroid_lon
}

print("Tâm của các tọa độ là:", centroid)
```
