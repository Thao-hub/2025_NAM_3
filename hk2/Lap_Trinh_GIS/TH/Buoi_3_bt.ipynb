```python
import random
cities_tw = [
    ("Hà Nội", 21.0285, 105.8542, "Bắc"),
    ("Hải Phòng", 20.8449, 106.6881, "Bắc"),
    ("Đà Nẵng", 16.0544, 108.2022, "Trung"),
    ("TP. Hồ Chí Minh", 10.8231, 106.6297, "Nam"),
    ("Cần Thơ", 10.0333, 105.7833, "Nam")
]
#E1: Các thành phố miền Bắc
for name, lat, lon, region in cities_tw:
    if region == "Bắc":
        print(f"Thành phố: {name}({lat}, {lon})")
```

```python
#E2: In tọa độ đến khi gặp vĩ độ < 11
index = 0
while index < len(cities_tw):
    name, lat, lon, region = cities_tw[index]
    print(f"{name}: ({lat}, {lon})")
    if lat < 11:
        break
    index += 1
```

```python
#E3: Thành phố miền Trung và miền Bắc
count = 0
for name, lat, lon, region in cities_tw:
    if region == "Trung" or region == "Bắc":
        count += 1
        print(f"STT {count}: {name} ({lat}, {lon})")
print(f"Tổng số thành phố miền Trung và miền Bắc: {count}")
```

```python
#E4: Tạo tọa độ ngẫu nhiên đến khi cả hai > 50
import random

rand_lat = random.uniform(0, 90)
rand_lon = random.uniform(0, 180)
print(f"Generated: ({rand_lat:.2f}, {rand_lon:.2f})")

while rand_lat > 50 :
    print(f"Generated: ({rand_lat:.2f}, {rand_lon:.2f})")
    rand_lat = random.uniform(0, 90)
    rand_lon = random.uniform(0, 180)
    
    
    if rand_lat > 50 and rand_lon > 50:
        continue
        print(f"Tọa độ 123: ({rand_lat:.2f}, {rand_lon:.2f})")
        
# continue / break loại bỏ tất cả code bên ngoài vòng lặp
```
